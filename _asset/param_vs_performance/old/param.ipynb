```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# 0. 폰트 설정 (기본 폰트 사용)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호는 유지

myblue="#5384EC"
myyellow="#F5B025"
myred ="#D85140"

# 1. 테이블 데이터 구조화
data = [
    {'Model': 'QHNet', 'Param': 21, 'H': 77.72, 'E_occ': 963.45, 'Color': myyellow}, # 주황색
    {'Model': 'WANet', 'Param': 25, 'H': 80.00, 'E_occ': 833.62, 'Color': 'C5'},
    {'Model': 'SPHNet', 'Param': 25, 'H': 45.48, 'E_occ': 334.28, 'Color': 'C4'},
    {'Model': 'QHFlow', 'Param': 28, 'H': 22.95, 'E_occ': 119.67, 'Color': myblue}, # 주황색
    {'Model': 'eSCFlow-small', 'Param': 12, 'H': 16.29, 'E_occ': 94.89, 'Color': myred}, # 파란색
    {'Model': 'eSCFlow-middle', 'Param': 43, 'H': 9.18, 'E_occ': 59.23, 'Color': myred}, # 파란색
    {'Model': 'eSCFlow-large', 'Param': 183, 'H': 5.93, 'E_occ': 45.73, 'Color': myred}, # 파란색
    {'Model': 'eSCFlow-xlarge', 'Param': 990, 'H': 4.58, 'E_occ': 39.36, 'Color': myred}, # 파란색
]

# 2. eSCFlow 라인용 데이터 준비
escflow_data = [d for d in data if 'eSCFlow' in d['Model']]
escflow_data_sorted = sorted(escflow_data, key=lambda x: x['Param'])
escflow_params = [d['Param'] for d in escflow_data_sorted]
escflow_h_errors = [d['H'] for d in escflow_data_sorted]


# --- 변경점 1: figsize=(4, 3)으로 수정 ---
fig, ax1 = plt.subplots(1, 1, figsize=(4, 2.5))

# 작은 사이즈에 맞게 폰트 크기 조절
label_fs = 9
annot_fs = 7 # 범례 폰트
tick_fs = 7


# --- 변경점 4: eSCFlow를 한 그룹으로, 나머지는 개별로 플롯 ---
# eSCFlow 점들만 수집
escflow_plot_params = [d['Param'] for d in escflow_data]
escflow_plot_errors = [d['H'] for d in escflow_data]

# 나머지 모델들 (개별 라벨)

other_models = [d for d in data if 'eSCFlow' not in d['Model']]
for d in other_models:
    ax1.scatter(d['Param'], d['H'], s=60, alpha=1.0, edgecolors=None, 
                zorder=2, color=d['Color'], label=d['Model'])

# eSCFlow 점들을 'eSCFlow'라는 단일 라벨로 플롯 (s=60으로 크기 조절)
ax1.scatter(escflow_plot_params, escflow_plot_errors, s=60, alpha=1.0, edgecolors=None, 
            zorder=2, color=myred, label='eSCFlow')

# --- 변경점 3: eSCFlow trend 라인을 label=None으로 그려 범례에서 제외 ---
ax1.plot(escflow_params, escflow_h_errors, linestyle='-', marker=None, 
         color=myred, alpha=1.0, zorder=1) # label=None (범례에서 제외)



# --- 스타일링 ---
# --- 변경점 2: set_title() 제거 ---

ax1.set_xlabel('Parameters (M)', fontsize=label_fs)
ax1.set_ylabel(r'$H \downarrow$ [$\mu E_h$]', fontsize=label_fs)
ax1.set_xscale('log')
ax1.set_yscale('log')

# --- 변경점 5: Minor grid 끄기 ---
ax1.grid(False, which='minor')
# --- 변경점 6: Major grid는 점선(':' = dots)으로 ---
ax1.grid(True, which='major', ls=':', alpha=1.0, zorder=0)

# --- *** 변경된 부분: Major Ticks 수동 설정 *** ---
# X축 (Parameters) 틱 위치 설정 (원하는 값으로 수정 가능)
x_ticks = [12,21,28, 43, 183, 990]
ax1.set_xticks(x_ticks)
ax1.set_xlim(10, 1100)

# Y축 (H error) 틱 위치 설정 (원하는 값으로 수정 가능)
y_ticks = [4.58, 5.93, 9.18,  16.29, 22.95, 77.77] 
ax1.set_yticks(y_ticks)
ax1.set_ylim(3.5, 100)

qhflow_h_error = 22.95
# ax1.axhline(y=qhflow_h_error, color=myblue, linestyle='-', linewidth=1.5)
ax1.text(50, qhflow_h_error+3, 'Previous SOTA', fontsize=7, color='k', ha='center', va='center')


# Ticks 포매터 (e.g., 10.0 -> 10)
ax1.xaxis.set_major_formatter(ticker.FuncFormatter(lambda y, _: '{:g}'.format(y)))
ax1.yaxis.set_major_formatter(ticker.FuncFormatter(lambda y, _: '{:g}'.format(y)))

# 틱은 안쪽으로
ax1.tick_params(direction='in', which='both', top=False, right=False, labelsize=tick_fs)
ax1.xaxis.set_minor_locator(ticker.NullLocator())
ax1.yaxis.set_minor_locator(ticker.NullLocator())

# 범례
ax1.legend(fontsize=annot_fs * 1.1, scatterpoints=1)

# 레이아웃
plt.tight_layout()
plt.show()

# # 5. (선택 사항) 이미지 파일로 저장
fig.savefig('model_performance_H_error_final.png', dpi=600, bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

# 0. 폰트 설정 (기본 폰트 사용)
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호는 유지

myblue = "#5384EC"
myyellow = "#F5B025"
myred = "#D85140"

# 1. 테이블 데이터 구조화
data = [
    {'Model': 'QHNet', 'Param': 20, 'H': 77.72, 'E_occ': 963.45, 'Color': myyellow, 'Marker': '^'}, # 원
    {'Model': 'WANet', 'Param': 28, 'H': 80.00, 'E_occ': 833.62, 'Color': 'C5', 'Marker': '^'}, # 사각형
    {'Model': 'SPHNet', 'Param': 25, 'H': 45.48, 'E_occ': 334.28, 'Color': 'C4', 'Marker': '^'}, # 삼각형
    {'Model': 'QHFlow', 'Param': 17, 'H': 27.95, 'E_occ': 119.67, 'Color': myblue, 'Marker': 'o'}, 
    {'Model': 'QHFlow', 'Param': 28, 'H': 22.95, 'E_occ': 119.67, 'Color': myblue, 'Marker': 'o'}, 
    {'Model': 'QHFlow', 'Param': 60, 'H': 18, 'E_occ': 119.67, 'Color': myblue, 'Marker': 'o'}, 
    {'Model': 'eSCFlow-small', 'Param': 12, 'H': 16.29, 'E_occ': 94.89, 'Color': myred, 'Marker': 'o'}, # 원
    {'Model': 'eSCFlow-25M', 'Param': 25, 'H': 11.76, 'E_occ': 70.44, 'Color': myred, 'Marker': 'o'}, # 사각형
    {'Model': 'eSCFlow-middle', 'Param': 43, 'H': 9.18, 'E_occ': 59.23, 'Color': myred, 'Marker': 'o'}, # 사각형
    {'Model': 'eSCFlow-large', 'Param': 183, 'H': 5.93, 'E_occ': 45.73, 'Color': myred, 'Marker': 'o'}, # 삼각형
    {'Model': 'eSCFlow-xlarge', 'Param': 990, 'H': 4.58, 'E_occ': 39.36, 'Color': myred, 'Marker': 'o'}, # 파란색
]

# 2. eSCFlow 라인용 데이터 준비
escflow_data = [d for d in data if 'eSCFlow' in d['Model']]
escflow_data_sorted = sorted(escflow_data, key=lambda x: x['Param'])
escflow_params = [d['Param'] for d in escflow_data_sorted]
escflow_h_errors = [d['H'] for d in escflow_data_sorted]

qhflow_data = [d for d in data if 'QHFlow' in d['Model']]
qhflow_params = [d['Param'] for d in qhflow_data]
qhflow_h_errors = [d['H'] for d in qhflow_data]


fig, ax1 = plt.subplots(1, 1, figsize=(4, 2.5))

label_fs = 9
annot_fs = 7 # 범례 폰트
tick_fs = 7

# --- eSCFlow 그룹 플롯 ---
# eSCFlow의 각 점은 다른 마커를 가짐 (개별 scatter 호출)
for d in escflow_data:
    ax1.scatter(d['Param'], d['H'], s=60, alpha=1.0, edgecolors=None, 
                zorder=2, color=d['Color'], marker=d['Marker'])

for d in qhflow_data:
    ax1.scatter(d['Param'], d['H'], s=60, alpha=1.0, edgecolors=None, 
                zorder=2, color=d['Color'], marker=d['Marker'])

# --- *** 변경점 1: eSCFlow trend 라인에 marker='o' 추가 (범례용) *** ---
# eSCFlow-small의 마커('o')를 대표 마커로 사용
ax1.plot(escflow_params, escflow_h_errors, linestyle='-', marker='o', markersize=7, 
         color=myred, alpha=1.0, zorder=1, label='eSCFlow')
ax1.plot(qhflow_params, qhflow_h_errors, linestyle='-', marker='o', markersize=7, 
         color=myblue, alpha=1.0, zorder=1, label='QHFlow')


# --- 나머지 모델들 (개별 라벨) ---
other_models = [d for d in data if 'eSCFlow' not in d['Model'] and 'QHFlow' not in d['Model']]
for d in other_models:
    ax1.scatter(d['Param'], d['H'], s=60, alpha=1.0, edgecolors=None, 
                zorder=2, color=d['Color'], label=d['Model'], marker=d['Marker'])


# --- 스타일링 ---
ax1.set_xlabel('Parameters', fontsize=label_fs)
ax1.set_ylabel(r'$H \downarrow$ [$\mu E_h$]', fontsize=label_fs)
ax1.set_xscale('log')
ax1.set_yscale('log')

# X축(세로) 그리드는 끄고, Y축(가로) 그리드만 켬
ax1.grid(False, which='minor')
# ax1.xaxis.grid(False) 
ax1.xaxis.grid(True, which='major', ls=':', alpha=1.0, zorder=0) 
ax1.yaxis.grid(True, which='major', ls=':', alpha=1.0, zorder=0) 


# --- *** 변경점 2: 모든 데이터 값을 Ticks로 설정 *** ---
# X축 (Parameters) 틱 위치 설정
# set으로 중복 제거 후 sorted로 정렬
# x_ticks = sorted(list(set([d['Param'] for d in data])))
# x_ticks.pop(-4)
x_ticks = [12,17,28, 43, 60, 183, 990]
ax1.set_xticks(x_ticks)
ax1.set_xlim(10, 1100) # xlim은 유지

# Y축 (H error) 틱 위치 설정
# set으로 중복 제거 후 sorted로 정렬
# y_ticks = sorted(list(set([d['H'] for d in data])))
y_ticks = [4.58, 5.93, 9.18, 11.76,  16.29, 22.95, 27.95, 77.77] 
# y_ticks.pop(-2)
ax1.set_yticks(y_ticks)
ax1.set_ylim(4, 100) # ylim은 유지

# ---------------------------------------------------

qhflow_h_error = 22.95
# ax1.axhline(y=qhflow_h_error, color=myblue, linestyle='-', linewidth=1.0)
# ax1.text(28, qhflow_h_error + 5, 'Previous SOTA', fontsize=7, color='k', ha='center', va='center')


ax1.xaxis.set_major_formatter(ticker.FuncFormatter(lambda y, _: '{:g}M'.format(y)))
ax1.yaxis.set_major_formatter(ticker.FuncFormatter(lambda y, _: '{:.2f}'.format(y)))

ax1.tick_params(direction='in', which='both', top=False, right=False, labelsize=tick_fs)
ax1.xaxis.set_minor_locator(ticker.NullLocator())
ax1.yaxis.set_minor_locator(ticker.NullLocator())

# 범례
handles, labels = ax1.get_legend_handles_labels()
order = ['QHNet', 'WANet', 'SPHNet', 'QHFlow', 'eSCFlow']

# 라벨이 존재하는지 확인 후 순서 정렬
handles_ordered = []
labels_ordered = []
for lbl in order:
    if lbl in labels:
        idx = labels.index(lbl)
        handles_ordered.append(handles[idx])
        labels_ordered.append(labels[idx])

ax1.legend(handles_ordered, labels_ordered, fontsize=annot_fs*0.9, scatterpoints=1, framealpha=1)

# 레이아웃
plt.tight_layout()
plt.show()

# # 5. (선택 사항) 이미지 파일로 저장
fig.savefig('model_performance_H_error_all_ticks.png', dpi=600, bbox_inches='tight')
```
