```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터 입력
x = np.array([4, 12, 25, 43, 183])
plot_x = np.array([4, 12, 25, 43, 183])
y = np.array([50.21, 15.44, 12.27, 8.70, 5.25])

# 로그 변환 후 1차 회귀 (y = a * log(x) + b)
log_x = np.log(x)
a, b = np.polyfit(log_x, y, 1)

# 예측값 계산
y_pred = a * np.log(plot_x) + b

# 결과 출력
print(f"Fitted equation: y = {a:.4f} * log(x) + {b:.4f}")

# 시각화
plt.scatter(x, y, color='blue', label='Data')
plt.plot(plot_x, y_pred, color='red', label='Log Fit')
plt.xscale('log')
plt.xlabel('x (log scale)')
plt.ylabel('y')
plt.title('Log Fitting: y = a * log(x) + b')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
x = np.array([4, 12, 25, 43, 183])
plot_x = np.array([4, 12, 25, 43, 183])
y = np.array([50.21, 15.44, 12.27, 8.70, 5.25])

# 로그 변환 후 선형 회귀 (log(y) = log(A) + b * log(x))
log_x = np.log(x)
log_y = np.log(y)

b, log_A = np.polyfit(log_x, log_y, 1)
A = np.exp(log_A)

# 예측값 계산
y_pred = A * plot_x**b

x_pred = 990
y_val = A * x_pred**b
print(f"Fitted equation: y = ({a:.4f} * x)^{b:.4f}")
print(f"Predicted y(990) = {y_val:.4f}")

# 결과 출력
print(f"Fitted equation: y = {A:.4f} * x^{b:.4f}")

# 시각화
plt.scatter(x, y, color='blue', label='Data')
plt.plot(plot_x, y_pred, color='red', label=f'Fit: y={A:.2f}x^{b:.2f}')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Power-law Fit: y = A * x^b')
plt.legend()
plt.grid(True, which='both', ls='--')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
x = np.array([4, 12, 25, 43, 183])
y = np.array([50.21, 15.44, 12.27, 8.70, 5.25])

# 로그 변환 후 선형 피팅
log_x = np.log(x)
log_y = np.log(y)
b, c = np.polyfit(log_x, log_y, 1)

# a 계산 (a = exp(c / b))
a = np.exp(c / b)

# 예측값
x_pred = 990
y_pred = (a * x_pred) ** b

# 결과 출력
print(f"Fitted equation: y = ({a:.4f} * x)^{b:.4f}")
print(f"Predicted y(990) = {y_pred:.4f}")

# 시각화
x_fit = np.linspace(min(x), max(x_pred, x[-1]), 200)
y_fit = (a * x_fit) ** b

plt.scatter(x, y, color='blue', label='Data')
plt.plot(x_fit, y_fit, color='red', label=f'y = ({a:.2f}x)^{b:.2f}')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Fit: y = (a x)^b')
plt.legend()
plt.grid(True, which='both', ls='--')
plt.show()
```
