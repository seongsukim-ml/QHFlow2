```python
import numpy
```

```python
middle_x = [5,10,20]
middle_H_error = [2.67,2.025,1.648]
middle_E_error = [2.63, 0.0098 ,0.010415]
middle_F_error = [1.69, 0.51,0.377243]

baseline_E = 3.2
baseline_F = 5.4

small_x = [5,10,20]
small_H_error = [3.28,2.65,2.34]
small_E_error = [40.68,0.0578,0.0321]
small_F_error = [12.387759 ,0.568247,0.453276]

import matplotlib.pyplot as plt
import numpy as np

# Figure 1: H error
fig1, ax1 = plt.subplots(figsize=(8, 6))
ax1.plot(middle_x, middle_H_error, 'o-', label='Middle', linewidth=2, markersize=8)
ax1.plot(small_x, small_H_error, 's-', label='Small', linewidth=2, markersize=8)
ax1.set_xlabel('Batch Size', fontsize=12)
ax1.set_ylabel('H Error', fontsize=12)
ax1.set_title('H Error vs Batch Size', fontsize=14, fontweight='bold')
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.set_xticks([5, 10, 20])
plt.tight_layout()
plt.show()

# Figure 2: E error
fig2, ax2 = plt.subplots(figsize=(8, 6))
ax2.plot(middle_x, middle_E_error, 'o-', label='Middle', linewidth=2, markersize=8)
ax2.plot(small_x, small_E_error, 's-', label='Small', linewidth=2, markersize=8)
ax2.axhline(y=baseline_E, color='r', linestyle='--', label=f'Equiformer ({baseline_E})', alpha=0.7)
ax2.set_xlabel('Batch Size', fontsize=12)
ax2.set_ylabel('E Error', fontsize=12)
ax2.set_title('E Error vs Batch Size', fontsize=14, fontweight='bold')
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)
ax2.set_xticks([5, 10, 20])
ax2.set_yscale('log')  # Log scale for better visualization
plt.tight_layout()
plt.show()

# Figure 3: F error
fig3, ax3 = plt.subplots(figsize=(8, 6))
ax3.plot(middle_x, middle_F_error, 'o-', label='Middle', linewidth=2, markersize=8)
ax3.plot(small_x, small_F_error, 's-', label='Small', linewidth=2, markersize=8)
ax3.axhline(y=baseline_F, color='r', linestyle='--', label=f'Equiformer ({baseline_F})', alpha=0.7)
ax3.set_xlabel('Batch Size', fontsize=12)
ax3.set_ylabel('F Error', fontsize=12)
ax3.set_title('F Error vs Batch Size', fontsize=14, fontweight='bold')
ax3.legend(fontsize=11)
ax3.grid(True, alpha=0.3)
ax3.set_xticks([5, 10, 20])
ax3.set_yscale('log')  # Log scale for better visualization
plt.tight_layout()
plt.show()

# Figure 4: Combined comparison at batch size 20
fig4, ax4 = plt.subplots(figsize=(8, 6))
width = 0.35

# Create grouped bar chart for batch size 20 comparison
batch_idx = 2  # Index for batch size 20
categories = ['H Error', 'E Error', 'F Error']
middle_values = [middle_H_error[batch_idx], middle_E_error[batch_idx], middle_F_error[batch_idx]]
small_values = [small_H_error[batch_idx], small_E_error[batch_idx], small_F_error[batch_idx]]

x = np.arange(len(categories))
ax4.bar(x - width/2, middle_values, width, label='Middle (b=20)', alpha=0.8)
ax4.bar(x + width/2, small_values, width, label='Small (b=20)', alpha=0.8)
ax4.set_ylabel('Error Value', fontsize=12)
ax4.set_title('Error Comparison at Batch Size 20', fontsize=14, fontweight='bold')
ax4.set_xticks(x)
ax4.set_xticklabels(categories, fontsize=11)
ax4.legend(fontsize=11)
ax4.grid(True, alpha=0.3, axis='y')
ax4.set_yscale('log')  # Log scale for better comparison
plt.tight_layout()
plt.show()
```
