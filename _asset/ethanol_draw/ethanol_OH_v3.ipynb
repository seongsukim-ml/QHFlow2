```python
# DIN 폰트 설치 및 설정
import matplotlib.font_manager as fm
import matplotlib.pyplot as plt
import os
import numpy as np

# DIN 폰트 파일 경로
font_dir = '/root/25DFT/font'

# 폰트 디렉토리에서 모든 DIN 폰트 파일 찾기
font_files = []
if os.path.exists(font_dir):
    for file in os.listdir(font_dir):
        if file.lower().endswith(('.otf', '.ttf', '.ttc')) and 'din' in file.lower():
            font_path = os.path.join(font_dir, file)
            font_files.append(font_path)
            print(f"✓ 폰트 파일 발견: {file}")

print(f"\n총 {len(font_files)}개의 DIN 폰트 파일을 찾았습니다.\n")

# matplotlib에 폰트 직접 등록
registered_fonts = []
for font_path in font_files:
    try:
        # 폰트를 matplotlib에 직접 추가
        fm.fontManager.addfont(font_path)
        # 폰트 정보 가져오기
        font_prop = fm.FontProperties(fname=font_path)
        font_name = font_prop.get_name()
        registered_fonts.append((font_name, font_path))
        print(f"✓ 등록 성공: {font_name} ({os.path.basename(font_path)})")
    except Exception as e:
        print(f"✗ 등록 실패: {os.path.basename(font_path)} - {str(e)}")

# 등록된 폰트 확인
if registered_fonts:
    print(f"\n총 {len(registered_fonts)}개의 폰트가 등록되었습니다.")
    
    # matplotlib에서 인식된 DIN 폰트 확인
    all_fonts = sorted(set([f.name for f in fm.fontManager.ttflist]))
    din_fonts = [f for f in all_fonts if 'din' in f.lower()]
    
    print(f"\nmatplotlib에서 인식된 DIN 폰트:")
    for font in din_fonts:
        print(f"  - {font}")
    
    # DIN_2014 우선 사용, 없으면 DIN-Regular 사용
    din_font_props = [f for f in fm.fontManager.ttflist if 'DIN' in f.name or 'din' in f.name.lower()]
    if din_font_props:
        # DIN_2014 폰트 우선 찾기
        din_2014 = [f for f in din_font_props if '2014' in f.name and 'Bold' not in f.name and 'Italic' not in f.name and 'Demi' not in f.name and 'Narrow' not in f.name and 'Extra' not in f.name]
        if din_2014:
            din_font_name = din_2014[0].name
        else:
            # DIN_2014가 없으면 일반 DIN-Regular 사용
            din_regular = [f for f in din_font_props if 'Bold' not in f.name and 'Italic' not in f.name and 'Condensed' not in f.name and 'Exp' not in f.name and '2014' not in f.name]
            if din_regular:
                din_font_name = din_regular[0].name
            else:
                # Regular가 없으면 첫 번째 DIN 폰트 사용
                din_font_name = din_font_props[0].name
        
        plt.rcParams['font.family'] = din_font_name
        plt.rcParams['font.sans-serif'] = [din_font_name] + plt.rcParams['font.sans-serif']
        print(f"\n✓ DIN 폰트를 기본 폰트로 설정했습니다: {din_font_name}")
    else:
        # 등록은 되었지만 이름이 다른 경우
        din_font_name = registered_fonts[0][0]
        plt.rcParams['font.family'] = din_font_name
        plt.rcParams['font.sans-serif'] = [din_font_name] + plt.rcParams['font.sans-serif']
        print(f"\n✓ DIN 폰트를 기본 폰트로 설정했습니다: {din_font_name}")
else:
    # DIN이 없으면 대체 폰트 사용
    available_fonts = [f.name for f in fm.fontManager.ttflist]
    if 'Arial' in available_fonts:
        plt.rcParams['font.family'] = 'Arial'
        print("\n⚠ DIN 폰트를 찾을 수 없습니다. Arial 폰트를 사용합니다.")
    elif 'Helvetica' in available_fonts:
        plt.rcParams['font.family'] = 'Helvetica'
        print("\n⚠ DIN 폰트를 찾을 수 없습니다. Helvetica 폰트를 사용합니다.")
    else:
        plt.rcParams['font.family'] = 'DejaVu Sans'
        print("\n⚠ DIN 폰트를 찾을 수 없습니다. DejaVu Sans 폰트를 사용합니다.")

plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호는 유지

# Bold 폰트 확인
print("\n" + "="*50)
print("DIN Bold 폰트 확인 중...")
print("="*50)

# Bold 폰트 파일이 있는지 확인
bold_font_files = [f for f in font_files if 'bold' in os.path.basename(f).lower() and 'italic' not in os.path.basename(f).lower()]
if bold_font_files:
    print(f"✓ Bold 폰트 파일 발견: {[os.path.basename(f) for f in bold_font_files]}")
    bold_available = True
else:
    print("⚠ 별도의 Bold 폰트 파일이 없습니다. (synthetic bold 사용 가능)")
    bold_available = False

# Bold 폰트 이름 확인 (DIN_2014 Bold 우선)
din_font_props = [f for f in fm.fontManager.ttflist if 'DIN' in f.name or 'din' in f.name.lower()]
# DIN_2014 Bold 우선 찾기
din_2014_bold = [f for f in din_font_props if '2014' in f.name and 'Bold' in f.name and 'Italic' not in f.name]
bold_font_names = [f.name for f in din_2014_bold] if din_2014_bold else [f.name for f in din_font_props if 'Bold' in f.name and 'Italic' not in f.name]
if bold_font_names:
    print(f"✓ Bold 폰트 이름: {bold_font_names}")
else:
    print("⚠ Bold 폰트 이름이 인식되지 않았습니다. (fontweight='bold'로 synthetic bold 시도)")
    
# 사용 가능한 모든 DIN weight 확인
print(f"\n사용 가능한 DIN 폰트 weight:")
weight_variants = {}
for font_prop in din_font_props:
    name = font_prop.name
    if 'Italic' not in name:
        if 'Bold' in name:
            weight_variants['Bold'] = name
        elif 'Condensed' in name:
            if 'Bold' in name:
                weight_variants['Condensed Bold'] = name
            else:
                weight_variants['Condensed'] = name
        elif 'Exp' in name:
            if 'Bold' in name:
                weight_variants['Exp Bold'] = name
            else:
                weight_variants['Exp'] = name
        else:
            weight_variants['Regular'] = name
            
for weight, name in sorted(weight_variants.items()):
    print(f"  - {weight}: {name}")

# 테스트 플롯 생성 - Regular vs Bold 비교
print("\n" + "="*50)
print("DIN 폰트 테스트 플롯 생성 중 (Regular & Bold 비교)...")
print("="*50)

fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# 첫 번째 플롯: 기본 테스트
ax1 = axes[0]
x = np.linspace(0, 10, 100)
y = np.sin(x)

ax1.plot(x, y, linewidth=2, label='sin(x)')
ax1.set_xlabel('X Axis (X축) - Regular', fontsize=14)
ax1.set_ylabel('Y Axis (Y축) - Regular', fontsize=14)
ax1.set_title('DIN Font Test - Regular Weight', fontsize=16, fontweight='normal')
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=12)

current_font = plt.rcParams['font.family']
ax1.text(5, 0.5, f'Current Font: {current_font}\nRegular Weight (fontweight="normal")', 
        fontsize=12, bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))

# 두 번째 플롯: Bold 테스트
ax2 = axes[1]
ax2.plot(x, y, linewidth=2, label='sin(x)', color='C1')
ax2.set_xlabel('X Axis (X축) - Bold', fontsize=14, fontweight='bold')
ax2.set_ylabel('Y Axis (Y축) - Bold', fontsize=14, fontweight='bold')
ax2.set_title('DIN Font Test - Bold Weight', fontsize=16, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend(fontsize=12, prop={'weight': 'bold'})

# Bold 텍스트 추가
bold_text = f'Current Font: {current_font}\nBold Weight (fontweight="bold")'
ax2.text(5, 0.5, bold_text, 
        fontsize=12, fontweight='bold',
        bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.5))

plt.tight_layout()
plt.show()

# 추가 테스트: 다양한 fontweight 비교
print("\n" + "="*50)
print("다양한 Font Weight 비교 테스트...")
print("="*50)

fig, ax = plt.subplots(figsize=(12, 8))
ax.axis('off')

# 다양한 fontweight 테스트
weights = ['normal', 'bold', 'light', 'heavy', 'black']
weight_names = ['Normal', 'Bold', 'Light', 'Heavy', 'Black']
y_positions = np.linspace(0.9, 0.1, len(weights))

test_text = f"DIN Font Test - {current_font}"

for i, (weight, name) in enumerate(zip(weights, weight_names)):
    try:
        ax.text(0.1, y_positions[i], f'{name} ({weight}): {test_text}', 
               fontsize=18, fontweight=weight, 
               transform=ax.transAxes,
               bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))
        print(f"✓ {name} ({weight}): 테스트 완료")
    except Exception as e:
        ax.text(0.1, y_positions[i], f'{name} ({weight}): [ERROR - {str(e)}]', 
               fontsize=18, transform=ax.transAxes,
               bbox=dict(boxstyle='round', facecolor='red', alpha=0.3))
        print(f"✗ {name} ({weight}): 오류 - {str(e)}")

ax.set_title('DIN Font Weight Comparison Test', fontsize=20, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

# DIN 특정 weight 직접 테스트
if 'weight_variants' in locals() and weight_variants:
    print("\n" + "="*50)
    print("DIN 특정 Weight 직접 테스트...")
    print("="*50)
    
    # weight 이름과 파일 경로 매핑 생성
    weight_to_file = {}
    for f_path in font_files:
        fname = os.path.basename(f_path).lower()
        if 'italic' not in fname:
            if 'bold' in fname:
                if 'condensed' in fname and 'Condensed Bold' not in weight_to_file:
                    weight_to_file['Condensed Bold'] = f_path
                elif 'exp' in fname and 'Exp Bold' not in weight_to_file:
                    weight_to_file['Exp Bold'] = f_path
                elif 'Bold' not in weight_to_file:
                    weight_to_file['Bold'] = f_path
            elif 'condensed' in fname and 'Condensed' not in weight_to_file:
                weight_to_file['Condensed'] = f_path
            elif 'exp' in fname and 'Exp' not in weight_to_file:
                weight_to_file['Exp'] = f_path
            elif 'Regular' not in weight_to_file:
                weight_to_file['Regular'] = f_path
    
    fig, ax = plt.subplots(figsize=(14, 10))
    ax.axis('off')
    
    y_pos = 0.95
    test_text_base = "DIN Font - The quick brown fox jumps over the lazy dog"
    
    for weight_name, font_name in sorted(weight_variants.items()):
        try:
            # 파일 경로 찾기
            font_path = None
            if weight_name in weight_to_file:
                font_path = weight_to_file[weight_name]
            else:
                # registered_fonts에서 찾기
                for reg_name, reg_path in registered_fonts:
                    if font_name in reg_name:
                        font_path = reg_path
                        break
            
            if font_path and os.path.exists(font_path):
                font_prop = fm.FontProperties(fname=font_path)
                ax.text(0.05, y_pos, f'{weight_name}: {test_text_base}', 
                       fontsize=14, fontproperties=font_prop,
                       transform=ax.transAxes,
                       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.2))
                print(f"✓ {weight_name} ({font_name}): 테스트 완료")
            else:
                # 폰트 이름으로 직접 사용
                font_prop = fm.FontProperties(family=font_name)
                ax.text(0.05, y_pos, f'{weight_name}: {test_text_base}', 
                       fontsize=14, fontproperties=font_prop,
                       transform=ax.transAxes,
                       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.2))
                print(f"✓ {weight_name} ({font_name}): 테스트 완료 (이름으로)")
            y_pos -= 0.12
        except Exception as e:
            ax.text(0.05, y_pos, f'{weight_name}: [ERROR - {str(e)}]', 
                   fontsize=14, transform=ax.transAxes,
                   bbox=dict(boxstyle='round', facecolor='red', alpha=0.3))
            print(f"✗ {weight_name}: 오류 - {str(e)}")
            y_pos -= 0.12
    
    ax.set_title('DIN All Weight Variants Test', fontsize=20, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.show()

print(f"\n✓ 모든 테스트 플롯이 생성되었습니다.")
print(f"  현재 사용 중인 폰트: {plt.rcParams['font.family']}")
if 'bold_available' in locals():
    print(f"  Bold 폰트 사용 가능: {bold_available}")
```

```python


# 데이터 로드 및 확인
import numpy as np
import matplotlib.pyplot as plt
import os

HA2eV = 27.211396641308   

# 데이터 디렉토리 경로
data_dir = '/root/25DFT/QHFlow/_asset/ethanol_draw/data'

# 데이터 파일 로드
cycles = np.load(os.path.join(data_dir, 'ethanol_dissociation_curves_d2_dft-100_cycles.npy'))
distances = np.load(os.path.join(data_dir, 'ethanol_dissociation_curves_d2_dft-100_distances.npy'))
energies_dft = np.load(os.path.join(data_dir, 'ethanol_dissociation_curves_d2_dft-100_energies_matched.npy'))
energies_dft = energies_dft * HA2eV

# energies_scflow = np.load(os.path.join(data_dir, 'ethanol_scflow_energies.npy'))
# energies_scflow = energies_scflow * HA2eV

energies_scflow = np.load(os.path.join(data_dir, 'ethanol_scflow_energies_new.npy'))

# UKS-DFT-100 에너지 데이터 로드
energies_uks_dft_raw = np.load(os.path.join(data_dir, 'ethanol_UKS-DFT-100-spin2_energies.npy'))
# energies_uks_dft_raw = energies_uks_dft_raw * HA2eV
print(f"UKS-DFT 원본 데이터 shape: {energies_uks_dft_raw.shape}")

# UKS-DFT 전용 distances 로드
distances_uks_dft = np.load(os.path.join(data_dir, 'ethanol_dissociation_curves_d2_dft-100_distances.npy'))
print(f"UKS-DFT distances shape: {distances_uks_dft.shape}")

# UKS 데이터의 dimension 처리
# UKS 전용 distances와 길이를 맞춰야 함
if energies_uks_dft_raw.ndim > 1:
    # 2D 이상인 경우: 첫 번째 차원이 거리/인덱스일 가능성
    if energies_uks_dft_raw.shape[0] == len(distances_uks_dft):
        energies_uks_dft = energies_uks_dft_raw[:, 0] if energies_uks_dft_raw.shape[1] > 1 else energies_uks_dft_raw.flatten()
    elif energies_uks_dft_raw.shape[1] == len(distances_uks_dft):
        energies_uks_dft = energies_uks_dft_raw[0, :] if energies_uks_dft_raw.shape[0] > 1 else energies_uks_dft_raw.flatten()
    else:
        # 길이가 다른 경우: 처음 len(distances_uks_dft)개만 사용
        energies_uks_dft = energies_uks_dft_raw.flatten()[:len(distances_uks_dft)]
        print(f"⚠ UKS 에너지 데이터 길이가 distances와 다릅니다. 처음 {len(distances_uks_dft)}개만 사용합니다.")
else:
    # 1D인 경우
    if len(energies_uks_dft_raw) == len(distances_uks_dft):
        energies_uks_dft = energies_uks_dft_raw
    elif len(energies_uks_dft_raw) > len(distances_uks_dft):
        # 더 긴 경우: 처음 len(distances_uks_dft)개만 사용
        energies_uks_dft = energies_uks_dft_raw[:len(distances_uks_dft)]
        print(f"⚠ UKS 에너지 데이터가 더 깁니다. 처음 {len(distances_uks_dft)}개만 사용합니다.")
    else:
        # 더 짧은 경우: distances_uks_dft의 처음 len(energies_uks_dft_raw)개에만 매칭
        distances_uks_dft = distances_uks_dft[:len(energies_uks_dft_raw)]
        energies_uks_dft = energies_uks_dft_raw
        print(f"⚠ UKS 에너지 데이터가 더 짧습니다. {len(energies_uks_dft_raw)}개만 사용합니다.")
        print(f"⚠ UKS distances도 처음 {len(energies_uks_dft_raw)}개만 사용합니다.")

# 에너지 단위 변환이 필요한지 확인 (이미 eV 단위일 수도 있음)
# 만약 Hartree 단위라면: energies_uks_dft = energies_uks_dft * HA2eV
# 여기서는 일단 그대로 사용하고, 필요시 변환

# 데이터 정보 출력
print("\n데이터 파일 정보:")
print(f"  cycles shape: {cycles.shape}, dtype: {cycles.dtype}")
print(f"  distances shape: {distances.shape}, dtype: {distances.dtype}")
print(f"  energies_dft shape: {energies_dft.shape}, dtype: {energies_dft.dtype}")
print(f"  energies_scflow shape: {energies_scflow.shape}, dtype: {energies_scflow.dtype}")
print(f"  energies_uks_dft shape: {energies_uks_dft.shape}, dtype: {energies_uks_dft.dtype}")
print(f"  distances_uks_dft shape: {distances_uks_dft.shape}, dtype: {distances_uks_dft.dtype}")
print(f"\n거리 범위:")
print(f"  DFT/SCFlow: {distances.min():.3f} ~ {distances.max():.3f} Å")
print(f"  UKS-DFT: {distances_uks_dft.min():.3f} ~ {distances_uks_dft.max():.3f} Å")
print(f"\n에너지 범위:")
print(f"  DFT: {energies_dft.min():.3f} ~ {energies_dft.max():.3f} eV")
print(f"  SCFlow: {energies_scflow.min():.3f} ~ {energies_scflow.max():.3f} eV")
print(f"  UKS-DFT-100: {energies_uks_dft.min():.3f} ~ {energies_uks_dft.max():.3f}")
```

```python
# 분해 곡선 시각화
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 첫 번째 플롯: 거리 vs 에너지 (DFT vs SCFlow vs UKS-DFT 비교)
ax1 = axes[0]
ax1.plot(distances, energies_dft, 'o-', label='DFT-100', linewidth=2, markersize=4, alpha=0.7)
ax1.plot(distances, energies_scflow, 's-', label='SCFlow', linewidth=2, markersize=4, alpha=0.7)
ax1.plot(distances_uks_dft, energies_uks_dft, '^-', label='UKS-DFT-100', linewidth=2, markersize=4, alpha=0.7, color='purple')
ax1.set_xlabel('Distance (Å)', fontsize=14, fontweight='bold')
ax1.set_ylabel('Energy (eV)', fontsize=14, fontweight='bold')
ax1.set_title('Ethanol OH Dissociation Curves', fontsize=16, fontweight='bold')
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.tick_params(labelsize=12)

# 두 번째 플롯: 에너지 차이 (SCFlow - DFT, UKS-DFT - DFT)
# UKS-DFT와 DFT의 거리가 다를 수 있으므로, 같은 거리 범위에서만 비교
# 또는 UKS-DFT의 각 거리에 대해 가장 가까운 DFT 값을 찾아서 비교
ax2 = axes[1]
energy_diff = energies_scflow - energies_dft
ax2.plot(distances, energy_diff, 'r-o', linewidth=2, markersize=4, alpha=0.7, label='SCFlow - DFT')

# UKS-DFT와 DFT 비교: UKS-DFT의 각 거리에 대해 가장 가까운 DFT 값을 찾아서 차이 계산
# 또는 단순히 UKS-DFT만 플롯 (DFT와의 차이는 거리가 다를 수 있어서 계산하기 어려움)
# 여기서는 UKS-DFT의 상대 에너지 (최소값 기준)를 표시
min_energy_all = min(energies_dft.min(), energies_scflow.min(), energies_uks_dft.min())
energy_uks_rel = energies_uks_dft - min_energy_all
energy_dft_rel = energies_dft - min_energy_all
# UKS-DFT의 각 거리에 대해 가장 가까운 DFT 거리를 찾아서 차이 계산
energy_diff_uks = []
distances_uks_for_diff = []
for i, dist_uks in enumerate(distances_uks_dft):
    # 가장 가까운 DFT 거리 인덱스 찾기
    closest_idx = np.argmin(np.abs(distances - dist_uks))
    diff = energies_uks_dft[i] - energies_dft[closest_idx]
    energy_diff_uks.append(diff)
    distances_uks_for_diff.append(dist_uks)
energy_diff_uks = np.array(energy_diff_uks)
distances_uks_for_diff = np.array(distances_uks_for_diff)

ax2.plot(distances_uks_for_diff, energy_diff_uks, 'm-^', linewidth=2, markersize=4, alpha=0.7, label='UKS-DFT - DFT (nearest)')
ax2.axhline(y=0, color='k', linestyle='--', linewidth=1, alpha=0.5)
ax2.set_xlabel('Distance (Å)', fontsize=14, fontweight='bold')
ax2.set_ylabel('Energy Difference (eV)', fontsize=14, fontweight='bold')
ax2.set_title('Energy Difference: SCFlow & UKS-DFT vs DFT', fontsize=16, fontweight='bold')
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.tick_params(labelsize=12)

# 통계 정보 추가
mean_diff = np.mean(energy_diff)
std_diff = np.std(energy_diff)
max_diff = np.max(np.abs(energy_diff))
mean_diff_uks = np.mean(energy_diff_uks)
std_diff_uks = np.std(energy_diff_uks)
max_diff_uks = np.max(np.abs(energy_diff_uks))
ax2.text(0.05, 0.95, f'SCFlow - DFT:\n  Mean: {mean_diff:.4f} eV\n  Std: {std_diff:.4f} eV\n  Max |Diff|: {max_diff:.4f} eV\n\nUKS-DFT - DFT:\n  Mean: {mean_diff_uks:.4f} eV\n  Std: {std_diff_uks:.4f} eV\n  Max |Diff|: {max_diff_uks:.4f} eV',
         transform=ax2.transAxes, fontsize=10,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()

print(f"\n에너지 차이 통계:")
print(f"  SCFlow - DFT:")
print(f"    평균: {mean_diff:.6f} eV")
print(f"    표준편차: {std_diff:.6f} eV")
print(f"    최대 절댓값: {max_diff:.6f} eV")
print(f"  UKS-DFT - DFT:")
print(f"    평균: {mean_diff_uks:.6f} eV")
print(f"    표준편차: {std_diff_uks:.6f} eV")
print(f"    최대 절댓값: {max_diff_uks:.6f} eV")
```

```python
# 방법 3: 여러 subplot으로 나누어서 다른 범위 보여주기

light_blue = "#7BA3F0"  # 더 연한 파란색
light_yellow = "#F5B025"
light_red = "#E87A6B"   # 더 연한 빨간색

myblue = "#5384EC"
myyellow = "#F5B025"
myred = "#D85140"
mygreen = "#009354"

darkred = "#C70600"
teal = "#4E79A7"

# 색상 정의 (나중에 쉽게 변경 가능)
color_qhflow2 = myblue  # 파란색
color_dft = "#D55E00"     # 주황색
color_uks_dft = "#9C27B0"  # 보라색 (UKS-DFT)
color_failure = myred      # 수렴 실패 영역 색상

# OH 거리 영역 색상 설정 (쉽게 변경 가능)
oh_region_color = "#bdbdbd"      # 영역 색상 (예: 'gray', '#808080', 'lightgray', 'darkgray' 등)
oh_region_alpha = 0.4        # 투명도 (0.0 ~ 1.0, 값이 클수록 진함)

# Marker 크기 설정 (쉽게 변경 가능)
marker_size_full = 2         # Subplot 1 (전체 범위)의 marker 크기
marker_size_zoom = 3         # Subplot 2, 3, 4 (확대 범위)의 marker 크기
marker_size_failure = 30     # 수렴 실패 영역 scatter marker 크기

# 거리 영역 경계값 정의 (나중에 쉽게 변경 가능)
distance_short_max = 0.65   # Short Distance: 0 ~ 0.75 Å
distance_medium_max = 1.5   # Medium Distance: 0.75 ~ 2.0 Å
distance_long_max = 10.0     # Long Distance: 2.0 ~ 10.0 Å

# OH 거리 범위 정의 (회색 영역)
oh_distance_min = 0.8694  # 최소값
oh_distance_max = 1.1342  # 최대값

ratio = 1.4
fig, axes = plt.subplots(2, 2, figsize=(5.*ratio, 2.8*ratio))

# 데이터 정렬
sort_idx = np.argsort(distances)
distances_sorted = distances[sort_idx]
energies_dft_sorted = energies_dft[sort_idx]
energies_qhflow2_sorted = energies_scflow[sort_idx]

# UKS-DFT 데이터 정렬 (자체 distances 사용)
sort_idx_uks = np.argsort(distances_uks_dft)
distances_uks_sorted = distances_uks_dft[sort_idx_uks]
energies_uks_dft_sorted = energies_uks_dft[sort_idx_uks]

# 최소값 기준 상대 에너지
min_energy = min(energies_dft_sorted.min(), energies_qhflow2_sorted.min(), energies_uks_dft_sorted.min())
energies_qhflow2_rel = energies_qhflow2_sorted - min_energy
energies_dft_rel = energies_dft_sorted - min_energy
energies_uks_dft_rel = energies_uks_dft_sorted - min_energy

# 수렴 실패 영역 찾기 (DFT가 QHFlow2보다 급격히 큰 에너지를 가지거나, 에너지 변화율이 매우 큰 부분)
energy_diff_dft = np.diff(energies_dft_rel)
energy_diff_qhflow2 = np.diff(energies_qhflow2_rel)
# DFT가 QHFlow2보다 에너지가 훨씬 크거나, 변화율이 2.5배 이상 큰 경우
convergence_fail_mask = np.zeros(len(distances_sorted), dtype=bool)
# diff는 길이가 1 작으므로 인덱스 조정
for i in range(len(energy_diff_dft)):
    if (energies_dft_rel[i+1] > energies_qhflow2_rel[i+1] + 50) or \
       (np.abs(energy_diff_dft[i]) > np.abs(energy_diff_qhflow2[i]) * 2.5):
        convergence_fail_mask[i+1] = True

# Subplot 1: 전체 범위
ax1 = axes[0, 0]
# 회색 영역 추가 (먼저 그려서 뒤에 오도록)
ax1.axvspan(oh_distance_min, oh_distance_max, alpha=oh_region_alpha, color=oh_region_color, 
           zorder=0, label='OH Distance Range (Train)')
ax1.plot(distances_sorted, energies_qhflow2_rel, 'o-', color=color_qhflow2, label='QHFlow2', 
         linewidth=2, markersize=marker_size_full, alpha=0.8)
ax1.plot(distances_sorted, energies_dft_rel, 's-', color=color_dft, label='DFT', 
         linewidth=2, markersize=marker_size_full, alpha=0.7)
ax1.plot(distances_uks_sorted, energies_uks_dft_rel, '^-', color=color_uks_dft, label='UKS-DFT-100', 
         linewidth=2, markersize=marker_size_full, alpha=0.7)
# 수렴 실패 영역 표시
if np.any(convergence_fail_mask):
    fail_idx = np.where(convergence_fail_mask)[0]
    ax1.scatter(distances_sorted[fail_idx], energies_dft_rel[fail_idx], 
               color=color_failure, s=marker_size_failure, alpha=0.5, marker='X', zorder=5, label='DFT Failure')
ax1.set_xlabel('Distance (Å)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Relative Energy (eV)', fontsize=12, fontweight='bold')
ax1.set_title('Full Range (0-10 Å)', fontsize=14, fontweight='bold')
legend1 = ax1.legend(fontsize=8, loc='best', edgecolor='black')
# QHFlow2 레이블만 bold로 설정
for text in legend1.get_texts():
    if text.get_text() == 'QHFlow2':
        text.set_fontweight('bold')
ax1.grid(True, alpha=0.3, axis='y')  # 가로선만 표시 (세로선 제거)
ax1.set_xlim(0, 10)
ax1.set_ylim(-10, 200)

# Subplot 2: 짧은 거리
ax2 = axes[0, 1]
# 회색 영역 추가 (해당 범위 내에 있는 경우만)
if oh_distance_max > 0 and oh_distance_min < distance_short_max:
    ax2.axvspan(max(0, oh_distance_min), min(distance_short_max, oh_distance_max), 
               alpha=oh_region_alpha, color=oh_region_color, zorder=0)
# 전체 데이터를 플롯하되 xlim으로 범위만 제한 (끊김 방지)
ax2.plot(distances_sorted, energies_qhflow2_rel, 'o-', color=color_qhflow2, 
         linewidth=2.5, markersize=marker_size_zoom, alpha=0.9)
ax2.plot(distances_sorted, energies_dft_rel, 's-', color=color_dft, 
         linewidth=2, markersize=marker_size_zoom, alpha=0.7)
ax2.plot(distances_uks_sorted, energies_uks_dft_rel, '^-', color=color_uks_dft, 
         linewidth=2, markersize=marker_size_zoom, alpha=0.7)
# 수렴 실패 영역 표시 (해당 범위 내)
mask1 = (distances_sorted >= 0) & (distances_sorted <= distance_short_max)
fail_mask1 = mask1 & convergence_fail_mask
if np.any(fail_mask1):
    fail_idx1 = np.where(fail_mask1)[0]
    ax2.scatter(distances_sorted[fail_idx1], energies_dft_rel[fail_idx1], 
               color=color_failure, s=marker_size_failure, alpha=0.4, marker='X', zorder=5)
ax2.set_xlabel('Distance (Å)', fontsize=12, fontweight='bold')
ax2.set_ylabel('Relative Energy (eV)', fontsize=12, fontweight='bold')
ax2.set_title(f'Short Distance (0-{distance_short_max} Å)', fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')  # 가로선만 표시 (세로선 제거)
ax2.set_xlim(0, distance_short_max)

# ax3의 y축 레이블을 "10k" 형식으로 표시
from matplotlib.ticker import FuncFormatter
def format_func(value, tick_number):
    if value >= 1000:
        return f'{value/1000:.0f}k'
    else:
        return f'{value:.0f}'
ax2.yaxis.set_major_formatter(FuncFormatter(format_func))


# Subplot 3: 중간 거리
ax3 = axes[1, 0]
# 회색 영역 추가 (해당 범위 내에 있는 경우만)
if oh_distance_max > distance_short_max and oh_distance_min < distance_medium_max:
    ax3.axvspan(max(distance_short_max, oh_distance_min), min(distance_medium_max, oh_distance_max), 
               alpha=oh_region_alpha, color=oh_region_color, zorder=0)
# 전체 데이터를 플롯하되 xlim으로 범위만 제한 (끊김 방지)
ax3.plot(distances_sorted, energies_qhflow2_rel, 'o-', color=color_qhflow2, 
         linewidth=2.5, markersize=marker_size_zoom, alpha=0.9)
ax3.plot(distances_sorted, energies_dft_rel, 's-', color=color_dft, 
         linewidth=2, markersize=marker_size_zoom, alpha=0.7)
ax3.plot(distances_uks_sorted, energies_uks_dft_rel, '^-', color=color_uks_dft, 
         linewidth=2, markersize=marker_size_zoom, alpha=0.7)
# 수렴 실패 영역 표시 (해당 범위 내)
mask2 = (distances_sorted >= distance_short_max) & (distances_sorted <= distance_medium_max)
fail_mask2 = mask2 & convergence_fail_mask
if np.any(fail_mask2):
    fail_idx2 = np.where(fail_mask2)[0]
    ax3.scatter(distances_sorted[fail_idx2], energies_dft_rel[fail_idx2], 
               color=color_failure, s=marker_size_failure, alpha=0.5, marker='X', zorder=5)
ax3.set_xlabel('Distance (Å)', fontsize=12, fontweight='bold')
ax3.set_ylabel('Relative Energy (eV)', fontsize=12, fontweight='bold')
ax3.set_title(f'Medium Distance ({distance_short_max}-{distance_medium_max} Å)', fontsize=14, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y')  # 가로선만 표시 (세로선 제거)
ax3.set_xlim(distance_short_max, distance_medium_max)
ax3.set_ylim(-0.5, 10)

# Subplot 4: 긴 거리
ax4 = axes[1, 1]
# 회색 영역 추가 (해당 범위 내에 있는 경우만 - 이 subplot에는 없을 것으로 예상되지만 추가)
if oh_distance_max > distance_medium_max and oh_distance_min < distance_long_max:
    ax4.axvspan(max(distance_medium_max, oh_distance_min), min(distance_long_max, oh_distance_max), 
               alpha=oh_region_alpha, color=oh_region_color, zorder=0)
# 전체 데이터를 플롯하되 xlim으로 범위만 제한 (끊김 방지)
ax4.plot(distances_sorted, energies_qhflow2_rel, 'o-', color=color_qhflow2, 
         linewidth=2.5, markersize=marker_size_zoom, alpha=0.9)
ax4.plot(distances_sorted, energies_dft_rel, 's-', color=color_dft, 
         linewidth=2, markersize=marker_size_zoom, alpha=0.7)
ax4.plot(distances_uks_sorted, energies_uks_dft_rel, '^-', color=color_uks_dft, 
         linewidth=2, markersize=marker_size_zoom, alpha=0.7)
# 수렴 실패 영역 표시 (해당 범위 내)
mask3 = (distances_sorted >= distance_medium_max) & (distances_sorted <= distance_long_max)
fail_mask3 = mask3 & convergence_fail_mask
if np.any(fail_mask3):
    fail_idx3 = np.where(fail_mask3)[0]
    ax4.scatter(distances_sorted[fail_idx3], energies_dft_rel[fail_idx3], 
               color=color_failure, s=marker_size_failure, alpha=0.5, marker='X', zorder=5)
ax4.set_xlabel('Distance (Å)', fontsize=12, fontweight='bold')
ax4.set_ylabel('Relative Energy (eV)', fontsize=12, fontweight='bold')
ax4.set_title(f'Long Distance ({distance_medium_max}-{distance_long_max} Å)', fontsize=14, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')  # 가로선만 표시 (세로선 제거)
ax4.set_xlim(distance_medium_max, distance_long_max)

ax4.yaxis.set_ticks([0, 10, 20, 30])
ax4.set_ylim(-0.2, 30)

plt.tight_layout()
plt.show()

# 그림 저장
import os
from datetime import datetime

# 저장 디렉토리 생성
save_dir = '/root/25DFT/QHFlow/_asset/ethanol_draw/plots'
os.makedirs(save_dir, exist_ok=True)

# 파일명 생성 (타임스탬프 포함)
timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
filename = f'ethanol_dissociation_curves_{timestamp}.png'
filepath = os.path.join(save_dir, filename)

# 그림 저장
fig.savefig(filepath, dpi=300, bbox_inches='tight', facecolor='white')
print(f"✓ 그림이 저장되었습니다: {filepath}")

# 추가로 PDF로도 저장
filename_pdf = f'ethanol_dissociation_curves_{timestamp}.pdf'
filepath_pdf = os.path.join(save_dir, filename_pdf)
fig.savefig(filepath_pdf, bbox_inches='tight', facecolor='white', dpi=600, pad_inches=0.02)
print(f"✓ PDF가 저장되었습니다: {filepath_pdf}")
```
