```python
# Load the nablaDFT psi4wfn links file
file_path = '/root/25DFT/QHFlow/_data_scripts/nabla_dft/nablaDFT_psi4wfn_links.txt'

# Read the file
with open(file_path, 'r') as f:
    links = f.readlines()

# Remove whitespace/newlines
links = sorted([link.strip() for link in links if link.strip()])

print(f"Total number of links: {len(links)}")
print(f"\nFirst 5 links:")
for i, link in enumerate(links[:5], 1):
    print(f"{i}. {link}")
    
print(f"\nLast 5 links:")
for i, link in enumerate(links[-5:], len(links)-4):
    print(f"{i}. {link}")
```

```python
file_name_list = ["archive", "archive_small", "wfns_moses_conformers_archive"]
links_dict = {"archive": [], "archive_small": [], "wfns_moses_conformers_archive": []}

for link in links:
    for file_name in file_name_list:
        link_file_name = link.split("/")[-1]
        if link_file_name.startswith(file_name):
            if file_name == "archive":
                if link_file_name.startswith("archive_small"):
                    continue
            links_dict[file_name].append(link)
```

```python
cnts = 0
for file_name in file_name_list:
    print(f"Number of {file_name} links: {len(links_dict[file_name])}")
    cnts += len(links_dict[file_name])
    print(f"First 5 {file_name} links:")
    if file_name == "wfns_moses_conformers_archive":
        links_dict[file_name] = sorted(links_dict[file_name], key=lambda x: int(x.split("_")[-1].split(".")[0]))
    else:
        links_dict[file_name] = sorted(links_dict[file_name])
    for i, link in enumerate(links_dict[file_name][:5], 1):
        print(f"{i}. {link}")
    print()

print(f"Total number of links: {cnts}")
print(f"Full list length: {len(links)}")
```
