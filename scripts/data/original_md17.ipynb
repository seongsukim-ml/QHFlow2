```python
# cur_data = Napthalene[0]
cur_data = Salicylic_acid[0]
cur_pos = cur_data.pos
cur_z = cur_data.z
mol = pyscf.gto.Mole()
mol_conf = [[cur_z[i], cur_pos[i]] for i in range(len(cur_z))]
```

```python
mol.build(verbose=0, atom=mol_conf, basis="def2svp", unit="ang")
mf = pyscf.scf.RKS(mol)
mf = mf.density_fit(only_dfj=True, auxbasis="weigend")  # def2/J 계열
# mf.xc = "pbe+TC"
mf.xc = "pbe"
mf.grids.level = 4
# mf.conv_tol = 1e-9
mf.get_ovlp().shape
```

```python
# mf.xc = "pbe+TC"
```

```python
energy = mf.kernel()
print(mf.cycles)
print(energy)
print(energy * HA2KCALPM)
print(cur_data.energy)
```
